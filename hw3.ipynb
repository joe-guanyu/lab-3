X_1+X_2\sim N(\mu_1+\mu_2,\sigma_1^2+\sigma_2^2).

### Example 2

If $R$ follows a Rayleigh distribution with scale parameter $\sigma$, then $Y=R^2$ follows an exponential distribution with rate $\lambda=1/(2\sigma^2)$. Its PDF is
$$
f_Y(y)=
\begin{cases}
\dfrac{1}{2\sigma^2}e^{-y/(2\sigma^2)}, & y\geq 0,\\
0, & y<0.
\end{cases}
$$

### 2a)

If $X_1$ and $X_2$ are independent exponential random variables with the same rate $\lambda$, then $S=X_1+X_2$ follows a gamma distribution with shape 2 and rate $\lambda$. Its PDF is
$$
f_S(s)=
\begin{cases}
\lambda^2 s e^{-\lambda s}, & s\geq 0,\\
0, & s<0.
\end{cases}
$$


### 2b)

Let $Z=X/Y$, where $X$ and $Y$ are independent standard normal random variables. Their joint PDF is
$$
f_{X,Y}(x,y)=\frac{1}{2\pi}e^{-(x^2+y^2)/2}.
$$

For $y>0$, $X/Y\le z$ means $X\le zy$. For $y<0$, the inequality reverses. Therefore,
$$
F_Z(z)=
\int_0^\infty\int_{-\infty}^{zy}f_{X,Y}(x,y)\,dx\,dy
+
\int_{-\infty}^0\int_{zy}^{\infty}f_{X,Y}(x,y)\,dx\,dy.
$$

Differentiating with respect to $z$ gives
$$
\begin{aligned}
f_Z(z)
&=\int_0^\infty y f_{X,Y}(zy,y)\,dy
-\int_{-\infty}^0 y f_{X,Y}(zy,y)\,dy\\
&=\frac{1}{\pi}\int_0^\infty y e^{-(1+z^2)y^2/2}\,dy\\
&=\frac{1}{\pi(1+z^2)}.
\end{aligned}
$$

Thus, $Z$ follows a standard Cauchy distribution:
$$
\boxed{f_Z(z)=\frac{1}{\pi(1+z^2)},\qquad -\infty<z<\infty.}
$$


### 2c)

Assuming $X,Y\sim N(0,1)$ independently, the amplitude is
$$
R=|X+iY|=\sqrt{X^2+Y^2}.
$$

Using polar coordinates,
$$
f_R(r)=\int_0^{2\pi}\frac{1}{2\pi}e^{-r^2/2}r\,d\theta
=re^{-r^2/2},\qquad r\geq0.
$$

Thus, $R$ follows a Rayleigh distribution with PDF
$$
f_R(r)=
\begin{cases}
re^{-r^2/2}, & r\geq0,\\
0, & r<0.
\end{cases}
$$
